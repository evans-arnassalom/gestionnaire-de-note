```python
# Version 1

R101 = float(input("Note R101 : "))
R102 = float(input("Note R102 : "))
R103 = float(input("Note R103 : "))

coef_R101 = float(input("Coefficient R101 : "))
coef_R102 = float(input("Coefficient R102 : "))
coef_R103 = float(input("Coefficient R103 : "))

total = (
    R101 * coef_R101
    + R102 * coef_R102
    + R103 * coef_R103 )

coefficients = (
    coef_R101
    + coef_R102
    + coef_R103 )

UE1 = total / coefficients

print("Moyenne UE1 :", UE1, "/ 20")
```

```python
# Version 2

import numpy as np

# création de la liste de toutes les matière du semestre 1
matière = [f"R{i}" for i in range (101,116)]+[f"SAE{i}" for i in range(11,17)]

#listes des coeff des 3 UE
list_coef1=[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
list_coef2=[4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
list_coef3=[4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

# boucle pour récupérer la liste des notes
list_notes = [i for i in range(21)] #liste utilisée pour les tests 

# calcul de la moyenne des UE arrondie à 3 chiffres après la virgule (weight 3) 
def calcul_moy(list_coef):
    return round(np.average(list_notes, weights = list_coef),3)

print(f"Les moyennes sont {calcul_moy(list_coef1)} pour l'UE1 {calcul_moy(list_coef2)} pour l'UE2 {calcul_moy(list_coef3)} pour l'UE3")
```

```python
# Version 3

import numpy as np 
# création de la liste de toutes les matière du semestre 1 + dico
matière = [f"R{i}" for i in range (101,116)]+[f"SAE{i}" for i in range(11,17)]

# listes des coeff des 3 UE
list_coef1=[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
list_coef2=[4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
list_coef3=[4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

list_notes = [i for i in range(21)] #liste utilisée pour les tests 

# Création des dictionnaires (matière -> note ou coefficient)
def dico(liste):
    return {cle : valeur for cle, valeur in zip(matière, liste)}

dico_notes = dico(list_notes)
dico_coef1 = dico(list_coef1)
dico_coef2 = dico(list_coef2)
dico_coef3 = dico(list_coef3)

# Calcul de la moyenne pondérée "à la main" comme dans la partie 1, pour chaque UE

def moyenne_dico(dico_coef):
    total_coef = 0
    somme = 0
    for cle in dico_coef.keys(): # pour chaque cle, donc pour chaque matière
        total_coef += dico_coef[cle] # cumul des coefficients (valeurs récupérées dans le dico des coeff)
        somme += dico_coef[cle]*dico_notes[cle] # cumul des produits note*coeff avec les valeurs récupérées dans les dicos
    return round(somme/total_coef, 3) # moyenne avec 3 chiffres après la virgule

print(f"La moyenne de l'UE1 vaut {moyenne_dico(dico_coef1)}")
print(f"La moyenne de l'UE2 vaut {moyenne_dico(dico_coef2)}")
print(f"La moyenne de l'UE3 vaut {moyenne_dico(dico_coef3)}")

print(dico_notes)
```
